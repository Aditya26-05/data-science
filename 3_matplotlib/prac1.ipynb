```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.xlabel("year")
plt.ylabel("runs")
plt.title("sachin carrer")
plt.plot(years,runs)
plt.show()
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli,'ro--' ,label="Kohli")
plt.plot(years, sehwag, 'bs--', label="Sehwag")

plt.xlabel("Years")
plt.ylabel("Runs")

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
x=np.arange(len(sachin))
width=0.3
plt.bar(x-width,sachin, width=width,label="sachin")
plt.bar(x,sehwag, width=width,label="sehwag")
plt.bar(x+width,kohli ,width=width,label="kohli")
plt.legend()
plt.xticks(x,years)
```

```python
# Data
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
explode=[0,0,0,1]
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels,explode=explode,  wedgeprops={'edgecolor': "#742144", 'linewidth': 7, 'linestyle': '-.', 'fill':True},startangle=40,autopct=1.1)
plt.show()
```

```python
import matplotlib.pyplot as plt
 
days = [1, 2, 3, 4, 5, 6, 7]  # Days of the week
studying = [3, 4, 3, 5, 4, 3, 4]
playing = [2, 2, 1, 1, 2, 3, 2]
watching_tv = [2, 1, 2, 2, 1, 1, 1]
sleeping = [5, 5, 6, 5, 6, 5, 5]
 
labels = ['Studying', 'Playing', 'Watching TV', 'Sleeping']
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral']
plt.stackplot(days,studying,playing,watching_tv,sleeping,labels=labels,colors=colors)
plt.legend()
```
