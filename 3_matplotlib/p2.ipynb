```python
import numpy as np
import matplotlib.pyplot as plt

a = [12, 34, 54, 23]
b = [24, 34, 54, 35]
yrs = [1, 2, 3, 4]

w = np.arange(len(yrs))
width = 0.25

plt.bar(w - width, a, width=width, label="adi")
plt.bar(w , b, width=width, label="bauna")

plt.xticks(w, yrs)
plt.legend()
plt.show()
```
